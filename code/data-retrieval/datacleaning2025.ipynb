```python
import csv
import numpy as np
import math
def read_data(filename):
    f = open(filename,'r', encoding='utf-8')
    reader = csv.reader(f)
    raw_matrix = []
    for row in reader:
        temp = []
        for i in range(0,len(row)):
            try:
                temp.append(float(row[i]))
            except ValueError:
                temp.append(row[i])
        raw_matrix.append(temp)
    f.close()
    return raw_matrix


'''def dict_category(filename):
    f = open(filename,'r', encoding='cp1252')
    reader = csv.reader(f)
    product_dict = {}
    for row in reader:
        temp = []
        for i in range(4,len(row)):
            temp.append(str(row[i]))
        product_dict[str(row[1])] = temp
    f.close()
    return product_dict'''

def list_category(filename):
    f = open(filename,'r', encoding='cp1252')
    reader = csv.reader(f)
    product_list = []
    for row in reader:
        product_list.append(str(row[1]))
    f.close()
    return product_list

def read_countrycode(filename):
    f = open(filename,'r', encoding='latin-1')
    reader = csv.reader(f)
    raw_matrix = []
    for row in reader:
        temp = []
        for i in range(0,len(row)):
            try:
                temp.append(float(row[i]))
            except ValueError:
                temp.append(row[i])
        raw_matrix.append(temp)
    f.close()
    return raw_matrix


def dict_countrycode(filename):

    countrycode = read_countrycode(filename)
    countrycode_dict = {}
    for i in range(1,len(countrycode)):
        countrycode_dict[str(countrycode[i][0])] = countrycode[i][1]
    return countrycode_dict
```
