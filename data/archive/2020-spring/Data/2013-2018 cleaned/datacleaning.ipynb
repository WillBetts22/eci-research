```python
import csv
import numpy as np
import math
def read_data(filename):
    f = open(filename,'r', encoding='utf-8')
    reader = csv.reader(f)
    raw_matrix = []
    for row in reader:
        temp = []
        for i in range(0,len(row)):
            try:
                temp.append(float(row[i]))
            except ValueError:
                temp.append(row[i])
        raw_matrix.append(temp)  
    f.close()
    return raw_matrix


def dict_category(filename):
    f = open(filename,'r', encoding='cp1252')
    reader = csv.reader(f)
    product_dict = {}
    for row in reader:
        temp = []
        for i in range(4,len(row)):
            temp.append(str(row[i]))
        product_dict[str(row[1])] = temp
    f.close()
    return product_dict

def list_category(filename):
    f = open(filename,'r', encoding='cp1252')
    reader = csv.reader(f)
    product_list = []
    for row in reader:
        product_list.append(str(row[1]))
    f.close()
    return product_list

def read_countrycode(filename):
    f = open(filename,'r', encoding='latin-1')
    reader = csv.reader(f)
    raw_matrix = []
    for row in reader:
        temp = []
        for i in range(0,len(row)):
            try:
                temp.append(float(row[i]))
            except ValueError:
                temp.append(row[i])
        raw_matrix.append(temp)  
    f.close()
    return raw_matrix


def dict_countrycode(filename):

    countrycode = read_countrycode(filename)
    countrycode_dict = {}
    for i in range(1,len(countrycode)):
        countrycode_dict[str(countrycode[i][0])] = countrycode[i][1]
    return countrycode_dict
```

```python
#preprocess the data (delete + transpose)
rawdata=read_data("2014_mega.csv")
#codes have to be modified a bit depending on mega data
rawdata=np.delete(rawdata,[0],1)
rawdata=np.transpose(rawdata)
rawdata
```

```python
#read in the country code as a dictionary
countrycode_dict = dict_countrycode("Country_Code.csv")
```

```python

category_dict = dict_category("Concise Category.csv")
category_list = list_category("Concise Category.csv")
    
#Append to the new matrix
matrix = open("2014clean.csv",'w',newline='')
writer = csv.writer(matrix)
writer.writerow(category_list)
    
trace = 0
    
l1 = len(rawdata)
l2 = len(rawdata[1])
l3 = len(category_list)
commoditycode = rawdata[0]

for i in range (1,l1):
    product_row = []
    countrycode=rawdata[i][0]
    #Map country code to country names
    countryname = countrycode_dict[str(countrycode)]
    product_row.append(countryname)
    
    #map commodity code to product categories
    for j in range(1,l3):
        product = category_list[j]
        index = category_dict[product]
        temp = []
        for k in range(1,l2):
            #check first 1 or 2 digits of commodity codes
            
            if (len(commoditycode[k])==7 and commoditycode[k][0] in index):
                
                temp.append(rawdata[i][k])
            elif (len(commoditycode[k])==8 and commoditycode[k][0:2] in index):
                
                temp.append(rawdata[i][k])
                
        total = np.sum(np.asarray(temp).astype(np.float))
        product_row.append(total)
    writer.writerow(product_row)
    trace+=1
    #print(trace)
matrix.close()

#if __name__ == "__main__":
#    main()
```
