# P-2026-11-20-1: Does dropping commodity chapters change HS4 ECI the way it changes grouped ECI?

**Question** (from `proposals/2026-11-19.md` #2): 11-19 found that dropping the commodity columns from the lab's 18-group ECI raises its Spearman with log GDPpc (+0.150 to +0.184 for −Energy, 2019–2022) but **lowers** its agreement with the Growth Lab's published ECI in 2020–2022. Tonight we run the same test at **HS4**: drop the same HS chapters from repo HS4 ECI and score M1 (ATLAS ECI) and M2 (log GDPpc). If HS4 shows the same "income up, ATLAS down" pattern, the effect is general. If HS4 barely moves, the grouped result is a coarse-grid artefact.

**Why tonight:** `QUEUE.md` Q01–Q16 are all taken on `main`, and every open `claude/nightly-*` branch has its run merged to `main` already. Under PROTOCOL §1.4 the question comes from `proposals/`. 11-19 #2 is self-contained and doesn't modify `lib/`, unlike 11-18 #1 / 11-19 #3. It also tells us whether 11-19 #1 (resource correction) is worth doing at HS4 or only at the grouped level.

## TL;DR

- **Same direction, much smaller size: rule B gives "mixed" for all three mirror variants.** At HS4, dropping the commodity chapters also raises the income correlation and lowers agreement with ATLAS ECI, in **4 of 4** years 2019–2022 for H1, H2 and H3. But the mean income gain is only **0.044 / 0.043 / 0.051**, against **0.170 / 0.156 / 0.141** for the grouped twins on the same countries. That's **0.26–0.36×** the grouped effect, below the pre-stated ½ threshold for "general". It also isn't small enough for "artefact" (max |Δ| 0.063–0.064 > 0.05).
- **No HS4 drop passes rule A (0 of 4).** Each one loses ATLAS agreement in every year (2019: −0.031 to −0.040, and all four 2019 bootstrap CIs exclude 0, e.g. H1 [−0.080, −0.009]). The 2019 income gain is significant only for H3 (+0.047, CI [0.005, 0.097]). Base HS4 ECI already has Spearman **0.972** with ATLAS ECI and **0.808** with log GDPpc in 2019, against ATLAS ECI's own income ρ of **0.819** on the same sample.
- **At HS4 the drop moves resource exporters *away* from ATLAS.** For the 32 resource-heavy countries (share ≥ 0.40), the mean distance to the ATLAS rank rises from **6.5 to 10.7**, while the distance to the income rank falls only from **16.7 to 14.9**. Grouped V1 instead barely moved the ATLAS distance (19.8 → 19.3) and cut the income distance by 10.7. The pattern is the same as 11-19's: the drop buys income fit by trading away complexity, and HS4 has more complexity agreement to lose.
- **HS4 ECI is robust to the drop; grouped ECI isn't.** Spearman with its own base is **0.902–0.972** for HS4 drops, against **0.583–0.819** for grouped drops. O1 and O4 agree in **25/25** HS4 matrices (grouped: 14 of 20 disagree), and all 45 O4 ECIs correlate positively with H0.
- **For the lab:** don't drop commodity chapters at either level. The grouped instability (11-19) is mostly a coarse-grid problem, and moving to HS4 removes most of it. A small, general income-vs-complexity trade-off remains. If a resource adjustment is wanted, it should be an explicit control (11-19 #1), not a deletion. Report-country side effect at HS4: Vietnam falls from 58 → 77–85 (2019), and Canada rises from 45 → 30–31.

## Why this matters for the lab

Agenda §0 link 1 (ECI predicts growth) depends on ECI measuring *capabilities*, not a disguised income or resource signal. 11-19 showed that at the grouped level, one column decision (keep or drop Energy) moves the income correlation by up to 0.18. That's as large as the gap between grouped and HS4 ECI. Two readings are possible:

1. **General:** commodity products pull ECI away from income at any granularity. Then the right fix is an explicit resource correction (11-19 #1) for any ECI the lab adopts, including an HS4 one.
2. **Coarse-grid artefact:** with 18 columns, Energy is 1 of 18 dimensions and dominates the second eigenvector. With ~1,200 HS4 products, the ~30 commodity products shouldn't matter much. Then the lab's long-run fix (move to HS4, Q04 / Maitree) solves the problem by itself, and there's no need for a resource correction.

This decides which of the two the lab should invest in.

## Method (pre-stated before any HS4 variant was scored tonight)

**HS4 variants** (repo HS4; RCA recomputed on the remaining products; Atlas-style `RCA ≥ 1`; countries left with no RCA>1 product drop out; sorted SVD):
- **H0** base (all chapters except 98/99)
- **H1** − chapters 25–27 (mirrors grouped V1, *Energy Drilling and Mining* = 25, 26, 27)
- **H2** − 25–27, 70–71 (mirrors V2, + *Precious Stones*)
- **H3** − 25–27, 70–82 (mirrors V3, + *Metals* = 72–82)
- **HR** − 26, 27, 71 (the repo's resource-share definition; the literal 11-19 proposal)

The country set is fixed by H0 (the ≥ US$1B filter uses total exports *before* any drop), so a drop changes the products, not who qualifies.

**Grouped comparison:** V0–V3 exactly as 11-19 (lab rules: drop-any-zero, strict `>`), recomputed tonight with the same `eci_o4`.

**Orientation:** O4 for every matrix (PCI negative with ubiquity; ECI takes the consistent sign). We record O1 (diversity) and whether each O4 ECI correlates positively with H0.

**Primary metrics** on the **joint common sample** each year (countries scored by all 5 HS4 and all 4 grouped variants, with ATLAS ECI and WDI GDPpc), so grouped and HS4 effect sizes are measured on the same countries:
- **M1:** Spearman with ATLAS published ECI, same year.
- **M2:** Spearman with log GDPpc (PPP, constant).

**Decision rule A (is an HS4 drop "better"?)**, same as 11-19: beats H0 on both M1 and M2 in 2019 **and** in ≥ 2 of 2020–2022. 2019 paired country bootstrap (2,000 draws) for the differences.

**Decision rule B (general vs artefact)**, applied to the mirror variants H1–H3 against their grouped twins V1–V3:
- **General** if the HS4 mirror shows the grouped pattern (ΔM2 > 0 **and** ΔM1 < 0) in ≥ 3 of 4 years 2019–2022 **and** its mean ΔM2 is at least half the grouped twin's.
- **Artefact** if the HS4 mirror's |ΔM2| < 0.05 and |ΔM1| < 0.05 in all 4 years.
- Otherwise **mixed**.

**Secondary (pre-stated):** (a) Spearman of each variant with its own base (how far the drop moves ECI); (b) 2019 rank gain H0→H1 vs resource share, and the resource-heavy group's (share ≥ 0.40) distance to income vs ATLAS ranks, as in 11-19 D2/D3; (c) M1/M2 on the larger HS4-only common sample, as a robustness check; (d) the 7 Spring 2026 report countries. No growth pilot tonight: 11-19 and Q10 found the 2019→2023 pilot null for every measure, including ATLAS ECI.

## Setup

```python
import sys, time, warnings
T0 = time.time()
sys.path.insert(0, '../../lib')
import numpy as np, pandas as pd
import matplotlib; matplotlib.use('Agg')
import matplotlib.pyplot as plt
from pathlib import Path
from scipy import stats
import eci, research as R
warnings.filterwarnings('ignore')
pd.set_option('display.width', 170); pd.set_option('display.max_columns', 30); pd.set_option('display.precision', 3)
FIG = Path('figures'); FIG.mkdir(exist_ok=True)
YEARS = [2019, 2020, 2021, 2022, 2023]; CHK = [2019, 2020, 2021, 2022]
REPORT = R.REPORT_COUNTRIES

def ch(a, b): return [f'{c:02d}' for c in range(a, b + 1)]
HVARS = {'H0 base': [], 'H1 -25-27': ch(25, 27), 'H2 -25-27,70-71': ch(25, 27) + ch(70, 71),
         'H3 -25-27,70-82': ch(25, 27) + ch(70, 82), 'HR -26,27,71': ['26', '27', '71']}
GVARS = {'V0 base': [], 'V1 -Energy': ['Energy Drilling and Mining'],
         'V2 -Energy -Precious': ['Energy Drilling and Mining', 'Precious Stones'],
         'V3 -Energy -Precious -Metals': ['Energy Drilling and Mining', 'Precious Stones', 'Metals']}
HN, GN = list(HVARS), list(GVARS)
TWIN = dict(zip(HN[1:4], GN[1:4]))

def o4(M):
    """Sorted SVD of a binary M, O4 orientation. Returns z-scored ECI and diagnostics."""
    Mv = M.to_numpy(float); kc, kp = Mv.sum(1), Mv.sum(0)
    B = Mv / np.sqrt(kc[:, None] * kp[None, :])
    U, S, Vt = np.linalg.svd(B, full_matrices=False)
    k = U[:, 1] / np.sqrt(kc); q = Vt[1] / np.sqrt(kp)
    s4 = -np.sign(np.corrcoef(q, kp)[0, 1]); s1 = np.sign(np.corrcoef(k, kc)[0, 1])
    e = pd.Series(k * s4, index=M.index); e = (e - e.mean()) / e.std()
    return e, {'n': M.shape[0], 'products': M.shape[1], 'O4 sign': int(s4), 'O1 sign': int(s1),
               'O4 margin': abs(np.corrcoef(q, kp)[0, 1]), 'O1 margin': abs(np.corrcoef(k, kc)[0, 1]),
               'eigengap 2-3': S[1] ** 2 - S[2] ** 2}

def eci_hs4(X, drop):
    keep = [c for c in X.columns if c[:2] not in drop]
    return o4(eci.mcp(X[keep], strict=False))

def eci_grp(X):
    return o4(eci.mcp(eci.notebook_filter(X), strict=True))

XH = {y: R.repo_hs(y, 4) for y in YEARS}
G = {y: R.lab_groups_iso3(y, drop_eu=True) for y in YEARS}
RES = {y: R.resource_share(XH[y]) for y in YEARS}
AT = R.atlas_cy(); ATL = {y: AT.query('year == @y').set_index('iso3').eci for y in YEARS}
LNY = {y: np.log(R.gdppc(y, 'ppp')).dropna() for y in YEARS}
E, D = {}, []
for y in YEARS:
    for v, drop in HVARS.items():
        E[y, v], d = eci_hs4(XH[y], drop); d.update(year=y, level='HS4', variant=v,
            dropped_products=XH[y].shape[1] - sum(c[:2] not in drop for c in XH[y].columns)); D.append(d)
    for v, drop in GVARS.items():
        E[y, v], d = eci_grp(G[y].drop(columns=drop)); d.update(year=y, level='grouped', variant=v, dropped_products=len(drop)); D.append(d)
VN = HN + GN
COMMON, HCOMMON = {}, {}
for y in YEARS:
    s = set(ATL[y].index) & set(LNY[y].index)
    h = set(s)
    for v in HN: h &= set(E[y, v].index)
    for v in VN: s &= set(E[y, v].index)
    COMMON[y], HCOMMON[y] = sorted(s), sorted(h)
print('lib:', Path(eci.__file__).name, Path(R.__file__).name, '| numpy', np.__version__, '| pandas', pd.__version__)
print('HS4 countries x products (H0):', {y: XH[y].shape for y in YEARS})
print('Joint common sample:', {y: len(COMMON[y]) for y in YEARS})
print('HS4-only common sample:', {y: len(HCOMMON[y]) for y in YEARS})
```

## Results

### A. Matrices, orientation and how far each drop moves ECI

```python
DD = pd.DataFrame(D).set_index(['year', 'level', 'variant'])
base_of = {**{v: 'H0 base' for v in HN}, **{v: 'V0 base' for v in GN}}
DD['rho w/ own base'] = [R.spearman(E[y, v], E[y, base_of[v]])[0] for (y, l, v) in DD.index]
DD['rho w/ H0'] = [R.spearman(E[y, v], E[y, 'H0 base'])[0] for (y, l, v) in DD.index]
print(DD[['n', 'products', 'dropped_products', 'O4 sign', 'O1 sign', 'O4 margin', 'O1 margin', 'eigengap 2-3', 'rho w/ own base', 'rho w/ H0']].round(3).to_string())
print('\nO1 and O4 disagree:', {l: int((g['O4 sign'] != g['O1 sign']).sum()) for l, g in DD.groupby(level='level')}, 'of', {l: len(g) for l, g in DD.groupby(level='level')})
print('O4 ECI positive with H0 in', int((DD['rho w/ H0'] > 0).sum()), 'of', len(DD), 'matrices')
print('HS4 n identical across HS4 variants every year:', all(DD.xs((y, 'HS4'))['n'].nunique() == 1 for y in YEARS))
```

**Takeaway (A):** HS4 drops remove 55–240 of ~1,224 products, yet the HS4 ECI keeps Spearman 0.902–0.972 with its base. The grouped drops remove 1–3 of 18 columns and fall to 0.583–0.819. O1 and O4 agree on all 25 HS4 matrices (O1 margin ≥ 0.396), while the grouped O1 margin hits 0.003 again (2019 V1), as in 11-19. The HS4 country set is identical across variants each year (the $1B filter is applied before the drop).

### B. Primary metrics on the joint common sample

```python
def metrics(SAMPLE, names):
    rows = []
    for y in YEARS:
        c = SAMPLE[y]
        for v in names:
            rows.append({'year': y, 'level': 'HS4' if v in HN else 'grouped', 'variant': v, 'n': len(c),
                         'M1 rho ATLAS': stats.spearmanr(E[y, v][c], ATL[y][c])[0],
                         'M2 rho lnGDPpc': stats.spearmanr(E[y, v][c], LNY[y][c])[0]})
    P = pd.DataFrame(rows)
    for m in ['M1', 'M2']:
        col = [k for k in P.columns if k.startswith(m)][0]
        P[m + ' diff vs base'] = P[col] - P.groupby(['year', 'level'])[col].transform('first')
    return P
P = metrics(COMMON, VN)
print(P.round(3).to_string(index=False))
ref = pd.DataFrame([{'year': y, 'n': len(COMMON[y]), 'rho(ATLAS ECI, lnGDPpc)': stats.spearmanr(ATL[y][COMMON[y]], LNY[y][COMMON[y]])[0]} for y in YEARS])
print('\nCeiling on the same sample:'); print(ref.round(3).to_string(index=False))
P.round(4).to_csv('primary_metrics.csv', index=False)
```

**Takeaway (B):** on the same 94–109 countries, every HS4 drop lowers M1 (−0.022 to −0.063) and raises M2 (+0.032 to +0.065) in every year, 2023 included. The grouped drops do the same at roughly 3–4× the size (ΔM2 +0.134 to +0.200, ΔM1 −0.003 to −0.151). The HS4 base sits near the ceiling: M2 0.764–0.820 against ATLAS ECI's own 0.749–0.819. Every HS4 drop exceeds that reference on M2 in every year (e.g. H3 2019: 0.855 vs 0.819), but so does H0 itself in 2020–2023. So exceeding it isn't by itself proof of income-proxying. The evidence for that is the simultaneous ATLAS loss.

### C. Decision rule A (HS4 drops) and 2019 paired bootstrap

```python
rng = np.random.default_rng(20261120)
c = np.array(COMMON[2019]); B = 2000
idx = rng.integers(0, len(c), size=(B, len(c)))
def boot(v, base, target):
    a = E[2019, v][c].to_numpy(); b = E[2019, base][c].to_numpy(); t = target[c].to_numpy()
    out = np.array([stats.spearmanr(a[i], t[i])[0] - stats.spearmanr(b[i], t[i])[0] for i in idx])
    return np.percentile(out, [2.5, 97.5])
dec = []
for v in HN[1:] + GN[1:]:
    base = base_of[v]; s = P[P.variant == v].set_index('year')
    win = (s['M1 diff vs base'] > 0) & (s['M2 diff vs base'] > 0)
    lo1, hi1 = boot(v, base, ATL[2019]); lo2, hi2 = boot(v, base, LNY[2019])
    dec.append({'variant': v, 'M1 diff 2019': s.loc[2019, 'M1 diff vs base'], 'M1 95% CI': f'[{lo1:.3f}, {hi1:.3f}]',
                'M2 diff 2019': s.loc[2019, 'M2 diff vs base'], 'M2 95% CI': f'[{lo2:.3f}, {hi2:.3f}]',
                'wins both, of 2020-22': int(win[[2020, 2021, 2022]].sum()),
                'BETTER (rule A)': bool(win[2019] and win[[2020, 2021, 2022]].sum() >= 2)})
DEC = pd.DataFrame(dec); print(DEC.round(3).to_string(index=False))
```

**Takeaway (C):** rule A rejects all 4 HS4 drops (and, as in 11-19, all 3 grouped drops on tonight's joint sample). At HS4 the ATLAS loss is the *more* certain effect: all four 2019 M1 CIs exclude 0, while three of four M2 CIs include 0. The grouped 2019 M1 change for V1 is −0.003 here, versus +0.017 in 11-19. The joint sample is smaller (105 countries, which must also be in the HS4 set), so the sign of that small difference isn't stable.

### D. Decision rule B: general effect or coarse-grid artefact?

```python
rows = []
for h, g in TWIN.items():
    sh = P[P.variant == h].set_index('year').loc[CHK]; sg = P[P.variant == g].set_index('year').loc[CHK]
    pat_h = int(((sh['M2 diff vs base'] > 0) & (sh['M1 diff vs base'] < 0)).sum())
    pat_g = int(((sg['M2 diff vs base'] > 0) & (sg['M1 diff vs base'] < 0)).sum())
    mh, mg = sh['M2 diff vs base'].mean(), sg['M2 diff vs base'].mean()
    general = pat_h >= 3 and mh >= 0.5 * mg
    artefact = bool((sh['M2 diff vs base'].abs() < 0.05).all() and (sh['M1 diff vs base'].abs() < 0.05).all())
    rows.append({'HS4 mirror': h, 'grouped twin': g, 'pattern yrs HS4 (of 4)': pat_h, 'pattern yrs grouped (of 4)': pat_g,
                 'mean dM2 HS4': mh, 'mean dM2 grouped': mg, 'ratio HS4/grouped': mh / mg,
                 'mean dM1 HS4': sh['M1 diff vs base'].mean(), 'mean dM1 grouped': sg['M1 diff vs base'].mean(),
                 'max |dM1|,|dM2| HS4': max(sh['M1 diff vs base'].abs().max(), sh['M2 diff vs base'].abs().max()),
                 'verdict': 'general' if general else ('artefact' if artefact else 'mixed')})
RB = pd.DataFrame(rows); print(RB.round(3).to_string(index=False))
```

**Takeaway (D):** the verdict is **mixed** for H1, H2 and H3. The direction is general (4/4 years, matching the grouped twins), but the size isn't: the HS4/grouped ratio of mean ΔM2 is 0.261, 0.274 and 0.360, below the pre-stated 0.5. HS4 effects reach 0.063–0.064, above the 0.05 artefact bound. Most of the grouped effect comes from the coarse grid, and a smaller part (about a quarter to a third) is a general trade-off.

### F. Mechanism: who moves when chapters 25–27 go (2019), at HS4 vs grouped

```python
y = 2019; c = [k for k in COMMON[y] if k in RES[y].index]
out = {}
for lvl, b, v in [('HS4', 'H0 base', 'H1 -25-27'), ('grouped', 'V0 base', 'V1 -Energy')]:
    r0 = E[y, b][c].rank(ascending=False); r1 = E[y, v][c].rank(ascending=False)
    out[lvl] = pd.DataFrame({'rank base': r0, 'rank drop': r1, 'gain': r0 - r1, 'resource share': RES[y][c],
                             'rank ATLAS': ATL[y][c].rank(ascending=False), 'rank lnGDPpc': LNY[y][c].rank(ascending=False)}).astype(float)
hi = out['HS4']['resource share'] >= 0.4
rows = []
for lvl, MV in out.items():
    g = MV[hi]
    rows.append({'level': lvl, 'n': len(MV), 'n resource-heavy': int(hi.sum()),
                 'rho(gain, resource share)': stats.spearmanr(MV['gain'], MV['resource share'])[0],
                 'mean |gain| all': MV['gain'].abs().mean(), 'mean gain heavy': g['gain'].mean(), 'mean gain others': MV.loc[~hi, 'gain'].mean(),
                 'heavy |rk-ATLAS| base': (g['rank base'] - g['rank ATLAS']).abs().mean(), 'heavy |rk-ATLAS| drop': (g['rank drop'] - g['rank ATLAS']).abs().mean(),
                 'heavy |rk-lnY| base': (g['rank base'] - g['rank lnGDPpc']).abs().mean(), 'heavy |rk-lnY| drop': (g['rank drop'] - g['rank lnGDPpc']).abs().mean()})
print(pd.DataFrame(rows).round(3).T.to_string())
print('\nHS4 top 8 gainers H0 -> H1 (2019):')
print(out['HS4'].sort_values('gain', ascending=False).head(8).round(2).to_string())
```

**Takeaway (F):** at HS4, rank gains still track resource share (ρ 0.513, against 0.664 grouped), but the moves are smaller (mean |gain| 5.9 vs 13.5 ranks). Their direction relative to ATLAS is worse: resource-heavy countries move 4.2 ranks *farther* from their ATLAS rank (6.5 → 10.7). Top HS4 gainers include Congo (+40, from 84 to 44, away from both its ATLAS rank of 101 and its income rank of 89), Australia (+25), Angola, Russia, Norway and Kazakhstan. DR Congo (+23) has resource share 0.12 under the crude 26/27/71 definition. It's a copper exporter (ch. 74), the known blind spot.

## Caveats

- **The thresholds in rule B (½ ratio, 0.05 band) are judgement calls**, pre-stated but not derived. The ratios (0.26–0.36) and the max |Δ| (0.063–0.064) are reported so a reader can apply their own.
- **ATLAS ECI isn't ground truth.** It is itself HS4-based (mirror-reconciled, with population and export floors), so HS4 variants are mechanically closer to it than grouped ones. That's why tonight compares *changes* against each level's own base rather than levels across grids.
- **Chapter mirrors are approximate.** Grouped "Energy Drilling and Mining" = chapters 25–27 exactly, but the grouped data also carries D4's mapping and the overlapping High-complexity aggregate. Dropping chapters at HS4 removes products the grouped aggregate never contained separately.
- **RCA threshold differs by level:** HS4 uses Atlas-style `≥ 1`, and grouped uses the lab's strict `> 1`, as in prior nights. Each level is compared only with its own base.
- **Small, dependent samples:** 94–109 countries on the joint sample, and years aren't independent. The bootstrap covers 2019 only. 2023 is suspect (D1) and is excluded from both rules.
- **Nominal values and crude resource share** (26/27/71 misses refined copper ch. 74, so DR Congo and Zambia look non-resource).

## Helpful to know

- **HS4 ECI is close to ATLAS's published ECI already:** Spearman 0.965–0.976 across 2019–2023 on both samples, with repo HS4 (98/99 dropped, ≥ US$1B, RCA ≥ 1, sorted SVD). Don't bother tuning HS4 variants to raise this further.
- **At HS4, the diversity rule (O1) is safe:** 25/25 matrices agree with O4. O1's failures are specific to grouped matrices (14/20 tonight, 2nd night running).
- **Income ρ alone can't rank HS4 variants.** Base HS4 ECI already matches or beats ATLAS ECI's own income ρ (0.749–0.819) in every year, and the drops push M2 to 0.81–0.86 while losing ATLAS agreement. Always pair income with an income-free benchmark.
- The ≥ US$1B filter must be applied **before** dropping products, or the country set changes across variants. `repo_hs()` does this when you drop columns afterwards, as tonight.
