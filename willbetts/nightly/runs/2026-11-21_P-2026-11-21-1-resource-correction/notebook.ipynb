# P-2026-11-21-1: Can a resource correction that *keeps* the commodity products improve the lab's ECI?

**Question** (from `proposals/2026-11-19.md` #1, and its HS4 version in the open 11-20 PR, `proposals/2026-11-20.md` #2): 11-19 and 11-20 showed that **deleting** the commodity columns/chapters raises ECI's Spearman with log GDPpc but lowers agreement with the Growth Lab's published ECI. That held at both levels, and was 3–4× larger on the lab's 18 groups. Tonight we test the two corrections those runs proposed, which keep every product in the matrix: (1) **residualise** ECI on resource export share, in the spirit of the Atlas's "ECI + natural-resource controls", with two resource definitions; and (2) **re-base RCA**, which removes resource exports from the RCA *denominators* only. We score both at the grouped and HS4 levels, with the same metrics and decision rule as 11-19/11-20.

**Why tonight:** `QUEUE.md` Q01–Q16 are all taken on `main`, and every open `claude/nightly-*` branch has its run claimed (11-20 is the only unmerged one; it took 11-19 #2). Under PROTOCOL §1.4, tonight's question comes from `proposals/`. 11-19 #1 is the highest-value untaken idea that doesn't modify `lib/`. Both 11-19 and 11-20 concluded "if a resource adjustment is wanted, it should be an explicit control, not a deletion", so this run tests that recommendation.

## TL;DR

- **No resource correction passes the pre-stated rule (0 of 6: three corrections × two levels). None of the six raises agreement with ATLAS ECI in any year 2019–2022.** The correction that keeps the column is not a free lunch either.
- **Residualising ECI on resource share (C1, C2) is the worst option.** It cuts M1 (ATLAS agreement) by **−0.166 to −0.286** (grouped) and **−0.115 to −0.196** (HS4) over 2019–2022. Every 2019 bootstrap CI excludes 0 (e.g. HS4 C1 **[−0.220, −0.052]**). It barely moves income fit (ΔM2 −0.070 to +0.042 grouped, −0.012 to +0.036 HS4). The reason: published ATLAS ECI itself correlates **r = −0.468 to −0.548** with resource share (2019–2022), so removing that relation pushes resource exporters *above* their ATLAS rank. The heavy group's mean rank gap to ATLAS goes from **+5.2 → −17.4** (grouped) and **+1.0 → −16.9** (HS4), and the linear fit makes **Nigeria #1 and Congo #2 of 105** on grouped ECI in 2019.
- **Re-based RCA (C3) behaves like the deletion it was meant to replace.** Grouped C3: ΔM2 **+0.164** (2019 CI [0.023, 0.301]) vs **+0.182** for drop-Energy. ΔM1 is **−0.022** (CI [−0.121, 0.078]) vs −0.003, and it moves ECI *more* than the drop (Spearman with base 0.667–0.754 vs 0.725–0.819). At HS4, C3 gives about the same income gain as dropping 26/27/71 (+0.028 to +0.043 vs +0.032 to +0.048) and loses ATLAS agreement in every year (−0.013 to −0.034), though less than the drop (−0.029 to −0.056).
- **The growth pilot is null for everything** (2019→2023, n = 105, p = 0.224–0.945, incl. ATLAS ECI). As a built-in check, residualising on the same control leaves the p-value unchanged (0.945 grouped, 0.526 HS4; Frisch–Waugh). So a resource-residualised ECI can never add growth information beyond "ECI + resource share" in a regression.
- **For the lab:** keep ECI uncorrected at both levels, and handle resources where the Atlas does, as a **separate control in growth/income comparisons**, not inside the ECI. Together with 11-19/11-20, this closes the "fix resource exporters inside ECI" line. The lab's real fix for grouped ECI remains the move to HS4 (base HS4 already has M1 0.972–0.976 with ATLAS).

## Why this matters for the lab

Agenda §0 link 1 (ECI predicts growth) needs ECI to measure capabilities, not resource dependence. Resource-rich rich countries (Norway, UAE, Australia, Canada) get low grouped ECI, and the Spring 2026 Canada and UAE reports print those ranks (UAE 118/120 in 2019 on the 18 groups, per 11-19). 11-19 and 11-20 rejected *deleting* commodity products and both recommended an explicit control instead. If an explicit control can lift income fit without losing complexity agreement, the lab could fix the resource-exporter ranks without changing data or the template. If it can't, the lab should stop trying to fix resource exporters *inside* the ECI and handle them in the interpretation (the Atlas approach).

## Method (pre-stated before any correction was scored tonight)

**Base ECIs** (as 11-20): *grouped* = lab rules (drop any country with a zero, strict `RCA > 1`), 18 columns; *HS4* = repo HS4, `RCA ≥ 1`. Both use a sorted SVD and **O4** orientation.

**Resource shares** (from repo HS raw, ISO3): **RS** = share of exports in HS 26, 27, 71 (the repo definition); **RSX** = RS + refined non-ferrous metals ch. 74–81 (Maitree's findings, next step #2: RS misses refined copper).

**Corrections (candidates):**
- **C1** residualise: OLS residual of base ECI on `[1, RS]`, z-scored. The fit uses the evaluation sample and is refit in every bootstrap draw.
- **C2** residualise on `[1, RSX]`.
- **C3** re-based RCA: `RCA_cp = (X_cp / NR_c) / (X_p / NR_world)`, where `NR` = exports *excluding* resource products. Resource products stay in M; only the denominators change. Resource products: HS4 = chapters 26/27/71; grouped = *Energy Drilling and Mining* + *Precious Stones* (ch. 25–27 and 70–71, the closest columns). Then the same binarisation, SVD and O4.

**References (not candidates):** the deletions from 11-19/11-20 (grouped −Energy, HS4 −26,27,71), recomputed on tonight's sample.

**Sample:** the **joint common sample** each year: countries scored by every spectral ECI at both levels, with ATLAS ECI, WDI GDPpc and RS.

**Primary metrics** (as 11-19/11-20): **M1** = Spearman with ATLAS published ECI, same year; **M2** = Spearman with log GDPpc (PPP constant).

**Decision rule A** (as 11-19/11-20; equivalent to 11-20 #2's "an M2 gain counts only if M1 doesn't fall"): a correction is *better* than its base only if it beats the base on **both** M1 and M2 in 2019 **and** in ≥ 2 of 2020–2022. 2019 paired country bootstrap (2,000 draws) for the differences.

**Secondary (pre-stated):** (S1) Spearman with ATLAS ECI *residualised on RS*, the like-for-like benchmark for C1. This is partly mechanical for C1/C2, so it can't override rule A. (S2) Spearman of each variant with its own base. (S3) Resource-heavy group (RS ≥ 0.40): mean |rank − ATLAS rank| and |rank − income rank|, as 11-19 D3. (S4) Growth pilot (Q10 spec), 2019→2023 growth ~ lnGDPpc + RS + z(ECI). (S5) The 7 Spring 2026 report countries.

## Setup

```python
import sys, time, warnings
T0 = time.time()
sys.path.insert(0, '../../lib')
import numpy as np, pandas as pd
import matplotlib; matplotlib.use('Agg')
import matplotlib.pyplot as plt
import statsmodels.api as sm
from pathlib import Path
from scipy import stats
import eci, research as R
warnings.filterwarnings('ignore')
pd.set_option('display.width', 180); pd.set_option('display.max_columns', 30); pd.set_option('display.precision', 3)
FIG = Path('figures'); FIG.mkdir(exist_ok=True)
YEARS = [2019, 2020, 2021, 2022, 2023]; CHK = [2019, 2020, 2021, 2022]
REPORT = R.REPORT_COUNTRIES
RES_CH = ['26', '27', '71']                                   # repo resource definition (lib/research.resource_share)
RESX_CH = RES_CH + [f'{c}' for c in range(74, 82)]            # + refined non-ferrous metals 74-81 (Maitree next step #2)
GRP_RES_COLS = ['Energy Drilling and Mining', 'Precious Stones']   # grouped columns closest to 26/27/71 (= ch. 25-27, 70-71)

def o4(M):
    """Sorted SVD of a binary M, O4 orientation (PCI negative with ubiquity). z-scored ECI + diagnostics. Copied from 11-19/11-20."""
    Mv = M.to_numpy(float); kc, kp = Mv.sum(1), Mv.sum(0)
    B = Mv / np.sqrt(kc[:, None] * kp[None, :])
    U, S, Vt = np.linalg.svd(B, full_matrices=False)
    k = U[:, 1] / np.sqrt(kc); q = Vt[1] / np.sqrt(kp)
    s4 = -np.sign(np.corrcoef(q, kp)[0, 1]); s1 = np.sign(np.corrcoef(k, kc)[0, 1])
    e = pd.Series(k * s4, index=M.index); e = (e - e.mean()) / e.std()
    return e, {'n': M.shape[0], 'products': M.shape[1], 'O4 sign': int(s4), 'O1 sign': int(s1),
               'O4 margin': abs(np.corrcoef(q, kp)[0, 1]), 'eigengap 2-3': S[1] ** 2 - S[2] ** 2}

def binarize(Rm, strict):
    M = ((Rm > 1) if strict else (Rm >= 1)).astype(float)
    return M.loc[M.sum(axis=1) > 0, M.sum(axis=0) > 0]

def rca_rebased(X, res_cols):
    """RCA with resource products removed from the DENOMINATORS only (country total and world total).
    Resource products stay in M; non-resource RCAs no longer get diluted by a large resource export total."""
    nr = X.drop(columns=res_cols)
    tot_c = nr.sum(axis=1)
    X = X.loc[tot_c > 0]; tot_c = tot_c[tot_c > 0]
    world_share = X.sum(axis=0) / nr.loc[X.index].to_numpy().sum()
    return X.div(tot_c, axis=0) / world_share

def grp_M(X, rebase=False, drop=()):
    X = eci.notebook_filter(X.drop(columns=list(drop)))          # lab rules: drop-any-zero, strict >
    Rm = rca_rebased(X, GRP_RES_COLS) if rebase else eci.rca(X)
    return binarize(Rm, strict=True)

def hs_M(X, rebase=False, drop=()):
    if drop:
        X = X[[c for c in X.columns if c[:2] not in drop]]
    rc = [c for c in X.columns if c[:2] in RES_CH]
    Rm = rca_rebased(X, rc) if rebase else eci.rca(X)
    return binarize(Rm, strict=False)                           # Atlas-style >= 1, as 11-20

def share(X, chs):
    cols = [c for c in X.columns if c[:2] in chs]
    return X[cols].sum(axis=1) / X.sum(axis=1)

XH = {y: R.repo_hs(y, 4) for y in YEARS}
G = {y: R.lab_groups_iso3(y, drop_eu=True) for y in YEARS}
RS = {y: share(XH[y], RES_CH) for y in YEARS}
RSX = {y: share(XH[y], RESX_CH) for y in YEARS}
AT = R.atlas_cy(); ATL = {y: AT.query('year == @y').set_index('iso3').eci for y in YEARS}
LNY = {y: np.log(R.gdppc(y, 'ppp')).dropna() for y in YEARS}

# Spectral ECIs (before any residualisation)
BASE, D = {}, []
SPEC = {('grouped', 'base'): lambda y: grp_M(G[y]), ('grouped', 'C3 rebased RCA'): lambda y: grp_M(G[y], rebase=True),
        ('grouped', 'ref: drop Energy'): lambda y: grp_M(G[y], drop=['Energy Drilling and Mining']),
        ('HS4', 'base'): lambda y: hs_M(XH[y]), ('HS4', 'C3 rebased RCA'): lambda y: hs_M(XH[y], rebase=True),
        ('HS4', 'ref: drop 26,27,71'): lambda y: hs_M(XH[y], drop=RES_CH)}
for y in YEARS:
    for (lvl, v), f in SPEC.items():
        BASE[y, lvl, v], d = o4(f(y)); d.update(year=y, level=lvl, variant=v); D.append(d)

# Joint common sample: scored by every spectral ECI at both levels, with ATLAS ECI, WDI GDPpc and resource share
COMMON = {}
for y in YEARS:
    s = set(ATL[y].index) & set(LNY[y].index) & set(RS[y].dropna().index)
    for k in [k for k in BASE if k[0] == y]: s &= set(BASE[k].index)
    COMMON[y] = sorted(s)

def resid(e, x):
    """OLS residual of e on [1, x], z-scored (fit on the index passed in)."""
    b = np.polyfit(x, e, 1)
    r = e - np.polyval(b, x)
    return (r - r.mean()) / r.std()

LV = ['grouped', 'HS4']
VN = {'grouped': ['base', 'C1 resid on RS', 'C2 resid on RSX', 'C3 rebased RCA', 'ref: drop Energy'],
      'HS4': ['base', 'C1 resid on RS', 'C2 resid on RSX', 'C3 rebased RCA', 'ref: drop 26,27,71']}
def scores(y, idx):
    """All variants on country index idx (residualisation refit on idx)."""
    out = {}
    for lvl in LV:
        b = BASE[y, lvl, 'base'][idx]
        out[lvl, 'base'] = b
        out[lvl, 'C1 resid on RS'] = resid(b.to_numpy(), RS[y][idx].to_numpy())
        out[lvl, 'C2 resid on RSX'] = resid(b.to_numpy(), RSX[y][idx].to_numpy())
        out[lvl, VN[lvl][3]] = BASE[y, lvl, VN[lvl][3]][idx]
        out[lvl, VN[lvl][4]] = BASE[y, lvl, VN[lvl][4]][idx]
    return {k: np.asarray(v, float) for k, v in out.items()}

S = {y: scores(y, COMMON[y]) for y in YEARS}
print('lib:', Path(eci.__file__).name, Path(R.__file__).name, '| numpy', np.__version__, '| pandas', pd.__version__)
print('HS4 matrix (base):', {y: BASE[y, 'HS4', 'base'].shape[0] for y in YEARS}, ' grouped (base):', {y: BASE[y, 'grouped', 'base'].shape[0] for y in YEARS})
print('Joint common sample:', {y: len(COMMON[y]) for y in YEARS})
```

### C. Decision rule A and 2019 paired bootstrap

```python
rng = np.random.default_rng(20261121)
y = 2019; c = np.array(COMMON[y]); B = 2000
IDX = rng.integers(0, len(c), size=(B, len(c)))
boot = {}
for i in IDX:
    ci = c[i]
    # resample countries, refit residualisation on the draw; spearman on the draw
    sc = {}
    for lvl in LV:
        b = BASE[y, lvl, 'base'][ci].to_numpy()
        sc[lvl, 'base'] = b
        sc[lvl, 'C1 resid on RS'] = resid(b, RS[y][ci].to_numpy())
        sc[lvl, 'C2 resid on RSX'] = resid(b, RSX[y][ci].to_numpy())
        sc[lvl, 'C3 rebased RCA'] = BASE[y, lvl, 'C3 rebased RCA'][ci].to_numpy()
    at = ATL[y][ci].to_numpy(); ly = LNY[y][ci].to_numpy()
    for lvl in LV:
        r0a, r0y = stats.spearmanr(sc[lvl, 'base'], at)[0], stats.spearmanr(sc[lvl, 'base'], ly)[0]
        for v in VN[lvl][1:4]:
            boot.setdefault((lvl, v), []).append((stats.spearmanr(sc[lvl, v], at)[0] - r0a, stats.spearmanr(sc[lvl, v], ly)[0] - r0y))
dec = []
for lvl in LV:
    for v in VN[lvl][1:4]:
        s = P[(P.level == lvl) & (P.variant == v)].set_index('year')
        win = (s['M1 diff'] > 0) & (s['M2 diff'] > 0)
        bb = np.array(boot[lvl, v]); q1 = np.percentile(bb[:, 0], [2.5, 97.5]); q2 = np.percentile(bb[:, 1], [2.5, 97.5])
        dec.append({'level': lvl, 'variant': v, 'M1 diff 2019': s.loc[2019, 'M1 diff'], 'M1 95% CI': f'[{q1[0]:.3f}, {q1[1]:.3f}]',
                    'M2 diff 2019': s.loc[2019, 'M2 diff'], 'M2 95% CI': f'[{q2[0]:.3f}, {q2[1]:.3f}]',
                    'M1 wins 2019-22': int((s.loc[2019:2022, 'M1 diff'] > 0).sum()), 'M2 wins 2019-22': int((s.loc[2019:2022, 'M2 diff'] > 0).sum()),
                    'both 2019': bool(win[2019]), 'both, of 2020-22': int(win[[2020, 2021, 2022]].sum()),
                    'BETTER (rule A)': bool(win[2019] and win[[2020, 2021, 2022]].sum() >= 2)})
DEC = pd.DataFrame(dec); print(DEC.round(3).to_string(index=False))
```

**Takeaway (C):** **0 of 6 corrections is "better".** None wins M1 in any year 2019–2022. Residualisation's 2019 M1 losses are all distinguishable from 0 (CIs from [−0.322, −0.103] to [−0.220, −0.052]), and none of its 2019 M2 changes is. C3's 2019 M2 gain is distinguishable from 0 at both levels (grouped [0.023, 0.301], HS4 [0.010, 0.070]), but its M1 loss is not (grouped [−0.121, 0.078], HS4 [−0.040, 0.002]). That is the same "income up, complexity flat-to-down" pattern 11-19 found for the deletion.

### D. Mechanism: where do resource-heavy countries move?

```python
# Who moves? Resource-heavy (RS >= 0.40) countries: mean |rank - ATLAS rank| and |rank - income rank|, 2019
y = 2019; c = COMMON[y]
rk = lambda a: pd.Series(a, index=c).rank(ascending=False)
ra, ry = rk(ATL[y][c].to_numpy()), rk(LNY[y][c].to_numpy())
hi = (RS[y][c] >= 0.40).to_numpy()
rows = []
for lvl in LV:
    for v in VN[lvl]:
        r = rk(S[y][lvl, v])
        rows.append({'level': lvl, 'variant': v,
                     f'heavy (n={hi.sum()}) |rk-ATLAS|': (r - ra).abs()[hi].mean(), 'heavy |rk-income|': (r - ry).abs()[hi].mean(),
                     f'others (n={(~hi).sum()}) |rk-ATLAS|': (r - ra).abs()[~hi].mean(), 'others |rk-income|': (r - ry).abs()[~hi].mean()})
print(pd.DataFrame(rows).round(1).to_string(index=False))
print('\n2019 resource-heavy countries in the common sample:', ', '.join(np.array(c)[hi]))
```

**Takeaway (D):** for the 32 resource-heavy countries (RS ≥ 0.40), residualisation roughly **doubles or triples** the distance to the ATLAS rank (grouped 19.8 → 30.1 / 29.1; HS4 6.5 → 17.4 / 19.4) and cuts the distance to the income rank only slightly (grouped 28.6 → 27.5 / 27.0; HS4 16.7 → 14.4 / 15.8). Its ATLAS distance also worsens for the other 73 countries (HS4 4.7 → 9.4 / 11.5). C3 behaves like the deletion: grouped heavy ATLAS distance 19.8 → 19.7 (drop 19.3), income distance 28.6 → 18.2 (drop 17.8).

```python
# Where do the heavy group's grouped ranks sit vs ATLAS before/after C1? (sign of the remaining gap)
y = 2019; c = COMMON[y]
T = pd.DataFrame({'RS': RS[y][c], 'rank ATLAS': rk(ATL[y][c].to_numpy()), 'rank lnY': rk(LNY[y][c].to_numpy())}, index=c)
for lvl in LV:
    for v in ['base', 'C1 resid on RS', 'C3 rebased RCA']:
        T[f'{lvl[:3]} {v.split()[0]}'] = rk(S[y][lvl, v])
H = T[T.RS >= 0.40].sort_values('RS', ascending=False)
print(H.round(2).to_string())
for lvl in LV:
    for v in ['base', 'C1 resid on RS', 'C3 rebased RCA']:
        col = f'{lvl[:3]} {v.split()[0]}'
        print(f'{col:12s}: heavy group mean (rank - ATLAS rank) = {(H[col] - H["rank ATLAS"]).mean():+.1f}  (positive = ranked below ATLAS)')
```

**Takeaway (D2):** at base, the heavy group sits close to its ATLAS rank on average (mean rank − ATLAS rank: grouped +5.2, HS4 +1.0). The corrections overshoot: grouped C1 −17.4 and C3 −14.6, HS4 C1 −16.9 and C3 −5.3 (negative = ranked *above* ATLAS). The linear residual is driven by the extreme-RS countries. On grouped C1, Nigeria (RS 0.87) ranks **1** and Congo (0.84) ranks **2** of 105, while ATLAS ranks them 104 and 101. Part of this comes from the grouped base, which already places Nigeria 20th (see New avenues #2).

### E. Growth pilot (Q10 spec), 2019 → 2023

```python
# Growth pilot (Q10 / 11-19 spec): 2019->2023 annualised real GDPpc growth ~ lnGDPpc2019 + RS + z(ECI variant), HC1
y = 2019; c = COMMON[y]
df = pd.DataFrame({'growth': R.growth(2019, 2023, 'const'), 'lny0': np.log(R.gdppc(2019, 'const'))}).reindex(c)
df['RS'] = RS[y][c]
names = []
for lvl in LV:
    for v in VN[lvl][:4]:
        nm = f'{lvl}: {v}'; df[nm] = S[y][lvl, v]; names.append(nm)
df['ATLAS ECI'] = (ATL[y][c] - ATL[y][c].mean()) / ATL[y][c].std(); names.append('ATLAS ECI')
core = df.dropna()
base = R.ols(core, 'growth', ['lny0', 'RS'])
rows = [{'model': 'baseline lnY0 + RS', 'n': base['n'], 'R2': base['r2']}]
for nm in names:
    m = R.ols(core.rename(columns={nm: 'x'}), 'growth', ['lny0', 'RS', 'x'])
    rows.append({'model': '+ ' + nm, 'n': m['n'], 'R2': m['r2'], 'coef per SD': m['coef']['x'], 'se': m['se']['x'], 'p': m['p']['x']})
print(pd.DataFrame(rows).round(4).to_string(index=False))
```

**Takeaway (E):** null across the board (all p ≥ 0.224; baseline R² 0.003). The C1 rows reproduce their base p-value exactly (grouped 0.945, HS4 0.526), only rescaled, as Frisch–Waugh predicts once RS is already a control. Residualising ECI on resource share can't add growth information beyond putting resource share in the regression. C3 is not significant at either level (grouped −0.340 per SD, p = 0.256; HS4 −0.103, p = 0.766).

## Caveats

- **ATLAS ECI is the benchmark, not ground truth**, and it carries a resource relation of its own (r ≈ −0.5 with RS here). A correction that is "right" about capabilities but disagrees with the Atlas would score badly on M1. S1 is meant to cover that case, but it is mechanically favourable to C1/C2, so it can't decide anything alone. Without an independent outcome (long-window growth, Q12/Q16), "ATLAS-like" is the best income-free yardstick in the repo.
- **Linear residualisation is the simplest control.** A rank-based or threshold control might overshoot less for extreme-RS countries. It still can't add growth information beyond the same control in a regression (section E).
- **The grouped C3 resource columns are imperfect:** *Energy Drilling and Mining* includes ch. 25 (salt, stone, cement), and *Precious Stones* includes ch. 70 (glass), per the lab mapping. HS4 C3 uses the exact 26/27/71 chapters.
- **Small, overlapping samples** (94–109 countries; the same countries each year). Bootstrap for 2019 only. Nominal values. D4's mapping is unchanged.
- **2023 is suspect (D1):** shown, but not used in the rule.
- **Growth pilot:** 4-year window including COVID. ATLAS ECI is also null, so this window can't rank measures.

## Helpful to know

- **Residualising ECI on a control that is already in the growth regression is a no-op for the test** (Frisch–Waugh, confirmed numerically in E: identical p-values). Any future "adjusted ECI" should be judged on outcomes that *don't* include that control.
- **Published ATLAS ECI correlates about r = −0.5 with resource share** (−0.468 to −0.548, 2019–2022, on these ~105 countries). The Atlas doesn't remove resource dependence from ECI; it controls for it in growth regressions. Any lab measure that zeroes this correlation will diverge from the Atlas by design.
- **Re-basing RCA denominators ≈ deleting the columns.** On 18 groups, they are statistically indistinguishable on M1/M2 (section B/C). Don't treat C3 as a "gentler" deletion at the grouped level.
- The grouped base ECI ranks **Nigeria 20th of 105 in 2019** (ATLAS: 104th). That's a grouped-level anomaly worth knowing about before any resource tweak on grouped data.
- `rca_rebased()` and the bootstrap with refit-in-draw residualisation (section C) are self-contained in this notebook and can be reused.
