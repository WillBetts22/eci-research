# P-2026-11-25-1: Should HS chapter 99 come out of the *High-complexity Products* aggregate?

**Question** (from `proposals/2026-11-24.md` #1, in open PR #20): the lab mapping puts chapter 99 (HS6 999999, "commodities not specified according to kind") in **both** *Services and Utilities* and the *High-complexity Products* aggregate. Saudi Arabia reports most of its oil under 999999 and ranks 2/105 on grouped ECI in 2019, against 45 on the Growth Lab's ECI. If ch. 99 comes out of the aggregate, alone (F4) or combined with 11-24's ch. 89 → Transportation fix (F1), does grouped ECI get better under the pre-stated rule A? And does it fix Saudi Arabia?

## TL;DR

- **No. Removing ch. 99 from the aggregate fails rule A, alone (F4) and combined with F1 (F5).** F4 improves both metrics in 2019 only (ΔM1 **+0.018** [−0.021, 0.073], ΔM2 **+0.009** [−0.031, 0.060]) and in **0 of 3** check years. It *lowers* the income fit in 2020, 2021 and 2022 (ΔM2 −0.026 / −0.016 / −0.046). F5 is no better than F1 alone in 2019 (ΔM1 +0.036 vs +0.037, ΔM2 +0.027 vs +0.034) and worse afterwards (0 of 3 check years). Removing ch. 99 entirely (F6: F4 plus dropping the Services column) also fails (2019 ΔM1 −0.002, ΔM2 −0.014). **F1 alone is still the only variant that passes** (recomputed tonight: +0.037 / +0.034, 3 of 3).
- **It only half-fixes Saudi Arabia, because the problem is in the data, not the mapping.** Under F4, Saudi Arabia goes from 4 → **13/105** in 2019 (ATLAS 45) and 4 → 11 in 2020. Saudi Arabia reports **$0.0B under ch. 27** and **$202.4B under 999999** in 2019 (same pattern in 2020 and 2023). In **2021** it reported **$212.6B under ch. 27** and $0.8B under 999999, and that year its grouped rank is **56/94**, close to ATLAS (44), with no mapping change at all. Taking ch. 99 out entirely (F6) deletes Saudi oil, so Saudi Arabia looks like a chemicals and plastics exporter instead (13/105 in 2019, 28/105 in 2020).
- **Most of F4's effect is a side effect, not the ch. 99 cells themselves.** Removing ch. 99 from the aggregate shrinks the world total over the 18 columns by **3.8%** ($21,414B → $20,593B), which moves every RCA denominator. In 2019, **23** M cells flip, **18 of them outside the aggregate column**, across 20 of 125 countries. The largest movers have almost no ch. 99 exports. Congo moves 41 → 85 because its *Daily Instruments* RCA moves 1.010 → 0.972. Moldova moves 24 → 46 (*Transportation* 1.037 → 0.998). Zambia moves 54 → 38 (*Precious Stones* 1.012 → 0.974).
- **For the reports: little changes, so there's no reason to adopt it.** Under F4 and F5, no Spring 2026 country changes its position among its three peers in 2019 or 2022. Poland gains the aggregate cell (RCA 0.964 → 1.077) and moves 33 → 21 (2019). Qatar, a UAE peer, is also a heavy 999999 reporter (85% of its exports in 2020, 77% in 2022), so the UAE peer sentences rest on the same data problem.

## Method (pre-stated before any variant was scored)

**Base:** the lab's grouped ECI rules (drop-any-zero filter, strict `RCA > 1`, sorted SVD, O4 orientation), on the **joint common sample** of 11-21/11-24 (cleaned grouped base ∩ HS4 base ∩ ATLAS ∩ WDI ∩ resource share: 105 countries in 2019). All variants are rebuilt from `data/raw` and compared with **R0** = the lab mapping rebuilt the same way.

**Variants:**
- **F4**: ch. 99 removed from *High-complexity Products* (it stays in *Services and Utilities*). This is the one-line change proposed in 11-24.
- **F5** = F1 + F4. F1 (ch. 89 *Daily Instruments* → *Transportation*) is the 11-24 variant, re-scored tonight as a reference.
- **F6**: ch. 99 out of the data entirely (F4, and the *Services and Utilities* column dropped). Under drop-any-zero, this also stops excluding countries whose only zero is ch. 99 (11-17). Comparisons stay on the same common sample.

**Metrics and rule** (unchanged from 11-19 to 11-24): **M1** = Spearman with ATLAS ECI, **M2** = Spearman with log GDPpc (PPP). **Rule A:** a variant is *better* only if it beats R0 on both M1 and M2 in 2019 **and** in ≥2 of 2020–2022. The 2019 paired country bootstrap uses 2,000 draws.

**Pre-stated diagnostics:**
- (A) Check that the cleaned aggregate really contains ch. 99, and count which countries' aggregate cell depends on it.
- (B3) Robustness: re-score excluding Saudi Arabia, and excluding every country whose aggregate cell flips under F4 that year.
- (C) Movers, and the cells that flip.
- (D) Spring 2026 report countries and their position among their 3 peers (peers from the `reports/2026-spring` PDFs, same dict as Q02/11-14/11-17).

*Added after the first look at the results (labelled as such):* C3 (why countries with no ch. 99 exports move) and C2b (Saudi Arabia's ch. 27 vs 999999 by year, plus a screen of all heavy 999999 reporters).

## Setup

```python
import sys, time, copy, warnings
T0 = time.time()
sys.path.insert(0, '../../lib')
import numpy as np, pandas as pd
import matplotlib; matplotlib.use('Agg')
import matplotlib.pyplot as plt
from pathlib import Path
from scipy import stats
import eci, research as R
warnings.filterwarnings('ignore')
pd.set_option('display.width', 200); pd.set_option('display.max_columns', 30); pd.set_option('display.precision', 3)
FIG = Path('figures'); FIG.mkdir(exist_ok=True)
YEARS = [2019, 2020, 2021, 2022, 2023]; CHK = [2019, 2020, 2021, 2022]
AGG, SVC = 'High-complexity Products', 'Services and Utilities'
PEERS = {'CAN': ['USA', 'AUS', 'GBR'], 'DEU': ['FRA', 'GBR', 'ITA'], 'POL': ['DEU', 'CZE', 'HUN'],
         'SGP': ['KOR', 'JPN', 'DEU'], 'ARE': ['SAU', 'KWT', 'QAT'], 'GBR': ['DEU', 'FRA', 'ITA'], 'VNM': ['SGP', 'MYS', 'PHL']}

def o4(M):
    """Sorted SVD of binary M, O4 orientation (PCI negative with ubiquity). Copied from 11-19/11-21/11-24.
    Returns z-scored ECI, raw k, raw q (PCI), s2."""
    Mv = M.to_numpy(float); kc, kp = Mv.sum(1), Mv.sum(0)
    B = Mv / np.sqrt(kc[:, None] * kp[None, :])
    U, S, Vt = np.linalg.svd(B, full_matrices=False)
    k = U[:, 1] / np.sqrt(kc); q = Vt[1] / np.sqrt(kp)
    s4 = -np.sign(np.corrcoef(q, kp)[0, 1]); k, q = k * s4, q * s4
    e = pd.Series(k, index=M.index); z = (e - e.mean()) / e.std()
    return z, e, pd.Series(q, index=M.columns), S[1]

def lab_M(X):
    """Lab rules: drop any country with a zero group, strict RCA > 1."""
    X = eci.notebook_filter(X)
    Mm = (eci.rca(X) > 1).astype(float)
    return Mm.loc[Mm.sum(1) > 0, Mm.sum(0) > 0]

def hs_M(X):
    Mm = (eci.rca(X) >= 1).astype(float)
    return Mm.loc[Mm.sum(1) > 0, Mm.sum(0) > 0]

def rank(e, idx):
    return e[idx].rank(ascending=False)

G = {y: R.lab_groups_iso3(y, drop_eu=True) for y in YEARS}
XH = {y: R.repo_hs(y, 4) for y in YEARS}
RS = {y: XH[y][[c for c in XH[y].columns if c[:2] in ['26', '27', '71']]].sum(1) / XH[y].sum(1) for y in YEARS}
AT = R.atlas_cy(); ATL = {y: AT.query('year == @y').set_index('iso3').eci for y in YEARS}
LNY = {y: np.log(R.gdppc(y, 'ppp')).dropna() for y in YEARS}
GB = {y: o4(lab_M(G[y])) for y in YEARS}; HB = {y: o4(hs_M(XH[y])) for y in YEARS}
COMMON = {y: sorted(set(GB[y][0].index) & set(HB[y][0].index) & set(ATL[y].dropna().index) & set(LNY[y].index) & set(RS[y].index)) for y in YEARS}
print('joint common sample n (same definition as 11-21/11-24):', {y: len(COMMON[y]) for y in YEARS})
print('SAU 2019 cleaned-base grouped rank:', int(rank(GB[2019][0], COMMON[2019])['SAU']), '/', len(COMMON[2019]),
      '| ATLAS rank:', int(rank(ATL[2019], COMMON[2019])['SAU']))
```

**Takeaway:** the sample matches 11-24 exactly (105 / 105 / 94 / 105 / 109). Saudi Arabia is 2/105 on the cleaned base in 2019, against 45 on ATLAS.

## Results

### A. Is ch. 99 in the aggregate, and whose aggregate cell depends on it?

```python
# A1. Is ch. 99 really inside the cleaned aggregate? Compare cleaned cell with raw sums with / without ch. 99.
L0 = R.LAB_GROUPS
print('lab mapping, aggregate chapters:', L0[AGG], '| Services chapters:', L0[SVC])
rows = []
for y in YEARS:
    raw = R.raw_hs6(y); ch = raw.index.str[:2].astype(int)
    agg_with = raw[np.isin(ch, L0[AGG])].sum(); agg_wo = raw[np.isin(ch, [c for c in L0[AGG] if c != 99])].sum()
    agg_with = agg_with.drop(index=[c for c in agg_with.index if c in eci.M49_AGGREGATES])
    agg_wo = agg_wo[agg_with.index]
    a = pd.DataFrame({'with': agg_with, 'wo': agg_wo}); a.index = [eci.m49_to_iso3(c) for c in a.index]
    a = a.loc[pd.notna(a.index)].groupby(level=0).sum()
    cl = G[y][AGG]; idx = sorted(set(cl.index) & set(a.index)); idx = [i for i in idx if cl[i] > 0]
    d_with = (np.abs(cl[idx] - a.loc[idx, 'with']) / cl[idx] < 1e-6).mean()
    d_wo = (np.abs(cl[idx] - a.loc[idx, 'wo']) / cl[idx] < 1e-6).mean()
    w99 = raw.loc[raw.index.str[:2] == '99'].drop(columns=[c for c in raw.columns if c in eci.M49_AGGREGATES]).to_numpy().sum()
    wagg = a['with'].sum()
    rows.append({'year': y, 'n countries': len(idx), 'share cleaned == raw WITH ch.99': round(d_with, 3),
                 'share cleaned == raw WITHOUT ch.99': round(d_wo, 3), 'world ch.99 $B': round(w99 / 1e9, 1),
                 'ch.99 share of world aggregate': round(a['with'].sum() - a['wo'].sum(), 0) / wagg if wagg else np.nan})
print(pd.DataFrame(rows).round(3).to_string(index=False))
```

**Takeaway (A1):** yes. The cleaned aggregate equals the raw chapter sum *with* ch. 99 for 90.0% of countries in 2019 and for 100% in 2022–2023 (16.9% and 12.8–14.3% without it). The 2019–2021 mismatches are the known cleaned-vs-raw differences (Q05). World ch. 99 is **$833.5B** in 2019, **14.2%** of the world aggregate column. Every dollar of it is counted twice, once in *Services and Utilities* and once in the aggregate.

```python
# A2. Which countries' aggregate RCA>1 depends on ch. 99? (lab rules, rebuilt R0 vs F4 = R0 minus ch.99 in aggregate)
F4 = copy.deepcopy(L0); F4[AGG] = [c for c in L0[AGG] if c != 99]
rows = []; FLIPS = {}
for y in YEARS:
    X0, X4 = R.groups_from_raw(y, L0), R.groups_from_raw(y, F4)
    X0f = eci.notebook_filter(X0); X4f = eci.notebook_filter(X4)
    idx = sorted(set(X0f.index) & set(X4f.index))
    r0, r4 = eci.rca(X0f).loc[idx, AGG], eci.rca(X4f).loc[idx, AGG]
    share99 = (X0f.loc[idx, AGG] - X4f.loc[idx, AGG]) / X0f.loc[idx, AGG]
    lost, gained = r0.index[(r0 > 1) & ~(r4 > 1)], r0.index[~(r0 > 1) & (r4 > 1)]
    FLIPS[y] = (list(lost), list(gained))
    rows.append({'year': y, 'n (lab filter)': len(idx), 'RCA>1 in agg, R0': int((r0 > 1).sum()), 'RCA>1 in agg, F4': int((r4 > 1).sum()),
                 'lose agg cell': ', '.join(lost), 'gain agg cell': ', '.join(gained),
                 'median ch.99 share of own agg cell': round(share99.median(), 4), 'n with ch.99 share >= 50%': int((share99 >= 0.5).sum())})
print(pd.DataFrame(rows).to_string(index=False))
y = 2019; X0f = eci.notebook_filter(R.groups_from_raw(y, L0)); X4f = eci.notebook_filter(R.groups_from_raw(y, F4))
s = ((X0f[AGG] - X4f[AGG]) / X0f[AGG]).sort_values(ascending=False).head(8)
print('\n2019 top ch.99 shares of the aggregate cell:')
print(pd.DataFrame({'ch.99 share of agg cell': s.round(3), 'agg cell $B (R0)': (X0f.loc[s.index, AGG] / 1e9).round(2),
                    'RCA agg R0': eci.rca(X0f).loc[s.index, AGG].round(2), 'RCA agg F4': eci.rca(X4f).loc[s.index, AGG].round(2)}).to_string())
```

**Takeaway (A2):** the aggregate cell is mostly ch. 99 for only a handful of countries: the median share is 5.1% in 2019, and 7 countries have a share ≥50%. Only **Saudi Arabia** has a large, *decisive* one (ch. 99 = **92.5%** of a $218.68B cell, RCA **1.73 → 0.25**). Russia's share is 71.6%, but its aggregate RCA is already below 1 (0.58). In 2019, F4 removes the aggregate cell from Denmark and Saudi Arabia and *adds* it for Korea, Poland and Romania, because everyone else's aggregate share rises once the $833B is gone from the world total.

### B. Rule A on the joint common sample

```python
# B. Variants, all rebuilt from data/raw and compared with R0 (the lab mapping rebuilt the same way).
F1 = copy.deepcopy(L0); F1['Daily Instruments'] = [c for c in L0['Daily Instruments'] if c != 89]; F1['Transportation'] = L0['Transportation'] + [89]
F5 = copy.deepcopy(F1); F5[AGG] = [c for c in L0[AGG] if c != 99]
F6 = copy.deepcopy(F4); del F6[SVC]
MAPS = {'R0 lab mapping (rebuilt)': L0, 'F1 ch.89 -> Transportation (11-24 ref.)': F1, 'F4 ch.99 out of aggregate': F4,
        'F5 = F1 + F4': F5, 'F6 ch.99 out entirely (F4, no Services col.)': F6}
VN = list(MAPS); V0 = VN[0]
VAR, PCI = {}, {}
for yy in YEARS:
    for name, mp in MAPS.items():
        z, k, q, s_ = o4(lab_M(R.groups_from_raw(yy, mp))); VAR[yy, name] = z; PCI[yy, name] = q
CS = {(yy, v): sorted(set(COMMON[yy]) & set(VAR[yy, V0].index) & set(VAR[yy, v].index)) for yy in YEARS for v in VN}
print('pairwise sample n:'); print(pd.DataFrame({v: [len(CS[yy, v]) for yy in YEARS] for v in VN}, index=YEARS).T.to_string())
print('grouped-matrix n (before common-sample restriction), 2019:', {v.split()[0]: len(VAR[2019, v]) for v in VN})
rows = []
for yy in YEARS:
    for v in VN:
        c = CS[yy, v]; e, e0 = VAR[yy, v][c], VAR[yy, V0][c]
        m1, m2 = stats.spearmanr(e, ATL[yy][c])[0], stats.spearmanr(e, LNY[yy][c])[0]
        rows.append({'year': yy, 'variant': v, 'n': len(c), 'M1 rho ATLAS': m1, 'M2 rho lnGDPpc': m2,
                     'M1 diff': m1 - stats.spearmanr(e0, ATL[yy][c])[0], 'M2 diff': m2 - stats.spearmanr(e0, LNY[yy][c])[0],
                     'rho w/ R0': stats.spearmanr(e, e0)[0],
                     'SAU rank': f"{int(rank(VAR[yy, v], c)['SAU'])}/{len(c)}" if 'SAU' in c else '-',
                     'SAU ATLAS': int(rank(ATL[yy], c)['SAU']) if 'SAU' in c else '-'})
P = pd.DataFrame(rows)
print(P.round(3).to_string(index=False))
P.round(4).to_csv(FIG / 'primary_metrics.csv', index=False)
print('R0 vs cleaned-CSV base, Spearman on the joint common sample:', {yy: round(stats.spearmanr(VAR[yy, V0][CS[yy, V0]], GB[yy][0][CS[yy, V0]])[0], 3) for yy in YEARS})
```

**Takeaway (B):** F4 raises both metrics only in 2019 (M1 0.689 → 0.707, M2 0.453 → 0.462). In 2020–2022 its M2 is below R0 every year (−0.026 / −0.016 / −0.046). F5 tracks F1 in 2019–2020, but it loses both metrics in 2021 (−0.022 / −0.029) and M2 in 2022 (−0.039), where F1 alone does not. F6 is negative on at least one metric every year. Saudi Arabia's rank: R0 4 → F4 13, F5 11, F6 13 (2019; ATLAS 45). The rebuilt R0 agrees with the cleaned base at Spearman 0.987–1.000. Saudi Arabia is **absent from the 2022 raw pull**, which is why its 2022 row shows "-".

```python
# B2. Rule A + 2019 paired country bootstrap (2,000 draws)
rng = np.random.default_rng(20261125); yy = 2019; B = 2000
dec = []
for v in VN[1:]:
    c = np.array(CS[yy, v]); at, ly = ATL[yy][c].to_numpy(), LNY[yy][c].to_numpy()
    e0, e1 = VAR[yy, V0][c].to_numpy(), VAR[yy, v][c].to_numpy()
    bb = []
    for i in rng.integers(0, len(c), size=(B, len(c))):
        bb.append((stats.spearmanr(e1[i], at[i])[0] - stats.spearmanr(e0[i], at[i])[0], stats.spearmanr(e1[i], ly[i])[0] - stats.spearmanr(e0[i], ly[i])[0]))
    bb = np.array(bb); s = P[P.variant == v].set_index('year')
    q1, q2 = np.percentile(bb[:, 0], [2.5, 97.5]), np.percentile(bb[:, 1], [2.5, 97.5])
    both = (s['M1 diff'] > 0) & (s['M2 diff'] > 0)
    dec.append({'variant': v, 'M1 diff 2019': round(s.loc[2019, 'M1 diff'], 3), 'M1 95% CI': f'[{q1[0]:.3f}, {q1[1]:.3f}]',
                'M2 diff 2019': round(s.loc[2019, 'M2 diff'], 3), 'M2 95% CI': f'[{q2[0]:.3f}, {q2[1]:.3f}]',
                'both 2019': bool(both[2019]), 'both, of 2020-22': int(both.loc[[2020, 2021, 2022]].sum()),
                'BETTER (rule A)': bool(both[2019] and both.loc[[2020, 2021, 2022]].sum() >= 2)})
DEC = pd.DataFrame(dec); print(DEC.to_string(index=False))
```

**Takeaway (B2):** **F4, F5 and F6 all fail rule A**, with 0 of 3 check years where both metrics improve. Every 2019 bootstrap interval for F4–F6 includes 0. F1 passes again, as in 11-24 (+0.037 [−0.002, 0.088] and +0.034 [0.004, 0.073]; the CIs differ slightly from 11-24 because the seed is new). Adding F4 to F1 never helps in 2019 (F5 +0.036 / +0.027 vs F1 +0.037 / +0.034).

```python
# B3. Pre-stated robustness: (i) excluding SAU; (ii) excluding every country that loses or gains its aggregate cell under F4 that year.
rows = []
for yy in CHK:
    movers = sorted(set(FLIPS[yy][0]) | set(FLIPS[yy][1]))
    for excl_name, excl in [('none', []), ('SAU', ['SAU']), ('F4 agg-cell flippers', movers)]:
        for v in VN[1:]:
            c = [i for i in CS[yy, v] if i not in excl]; e, e0 = VAR[yy, v][c], VAR[yy, V0][c]
            rows.append({'excluded': excl_name, 'year': yy, 'variant': v.split()[0], 'n': len(c),
                         'M1 diff': stats.spearmanr(e, ATL[yy][c])[0] - stats.spearmanr(e0, ATL[yy][c])[0],
                         'M2 diff': stats.spearmanr(e, LNY[yy][c])[0] - stats.spearmanr(e0, LNY[yy][c])[0]})
RB = pd.DataFrame(rows)
print(RB.pivot_table(index=['excluded', 'variant'], columns='year', values=['M1 diff', 'M2 diff'], sort=False).round(3).to_string())
```

**Takeaway (B3):** the failure isn't driven by Saudi Arabia or by the countries whose aggregate cell flips. Excluding Saudi Arabia, F4's ΔM2 is still −0.025 / −0.020 / −0.046 in 2020–2022. Excluding the flippers, it is −0.023 / −0.011 / −0.018. F1's pass survives both exclusions (ΔM1 and ΔM2 are positive in every year and in both exclusions).

### C. What moves, and why

```python
# C. What moves? Largest 2019 movers under F4, and group PCI ranks R0 vs F4
yy = 2019; c = CS[yy, VN[2]]; r0, r4 = rank(VAR[yy, V0], c), rank(VAR[yy, VN[2]], c)
mv = (r4 - r0).sort_values()
print('F4 2019 largest movers (negative = up):')
print(pd.DataFrame({'R0': r0[mv.index], 'F4': r4[mv.index], 'change': mv, 'ATLAS': rank(ATL[yy], c)[mv.index]}).iloc[np.r_[0:6, -6:0]].astype(int).to_string())
print('countries moving >= 10 ranks under F4:', {yy_: int(((rank(VAR[yy_, VN[2]], CS[yy_, VN[2]]) - rank(VAR[yy_, V0], CS[yy_, VN[2]])).abs() >= 10).sum()) for yy_ in CHK})
pr = pd.DataFrame({v.split()[0]: PCI[yy, v].rank(ascending=False) for v in [V0, VN[2], VN[4]]})
print('\n2019 group PCI rank (1 = most complex):'); print(pr.astype('Int64').sort_values('R0').to_string())
```

**Takeaway (C):** in 2019, 6 countries move ≥10 ranks under F4 (7 / 6 / 10 in 2020–2022). The largest movers are Congo (41 → 85), Moldova (24 → 46), Zambia (54 → 38), Denmark (15 → 29) and Poland (33 → 21). Saudi Arabia moves only 9 places. Group PCI ranks barely change: the aggregate stays 2nd of 18 under F4 and F6. So removing ch. 99 doesn't make the aggregate "less complex". It changes who holds it.

```python
# C3. Why do countries with little ch.99 (COG, MDA, ZMB, POL) move? Basket changes R0 -> F4 with the RCA of every changed cell, 2019.
yy = 2019
X0f, X4f = eci.notebook_filter(R.groups_from_raw(yy, L0)), eci.notebook_filter(R.groups_from_raw(yy, F4))
R0m, R4m = eci.rca(X0f), eci.rca(X4f)
w0, w4 = X0f.to_numpy().sum(), X4f.to_numpy().sum()
print(f'world total over the 18 columns (lab filter sample): R0 ${w0/1e9:,.0f}B, F4 ${w4/1e9:,.0f}B ({w4/w0-1:+.1%}); world aggregate column {X0f[AGG].sum()/1e9:,.0f}B -> {X4f[AGG].sum()/1e9:,.0f}B')
rows = []
for f in ['ZMB', 'POL', 'ROU', 'DNK', 'MDA', 'COG', 'SAU']:
    ch_ = [g for g in R0m.columns if (R0m.loc[f, g] > 1) != (R4m.loc[f, g] > 1)]
    rows.append({'iso3': f, 'ch.99 share of own total (R0 cols)': round((X0f.loc[f].sum() - X4f.loc[f].sum()) / X0f.loc[f].sum(), 3),
                 'cells that flip (RCA R0 -> F4)': '; '.join(f'{g} {R0m.loc[f, g]:.3f} -> {R4m.loc[f, g]:.3f}' for g in ch_)})
print(pd.DataFrame(rows).to_string(index=False))
n_flip = int(((R0m > 1) != (R4m > 1)).to_numpy().sum()); n_flip_non_agg = int(((R0m > 1) != (R4m > 1)).drop(columns=[AGG]).to_numpy().sum())
print(f'2019: M cells that flip under F4: {n_flip} (of which outside the aggregate column: {n_flip_non_agg}); countries with >=1 flip: {int(((R0m > 1) != (R4m > 1)).any(axis=1).sum())} of {len(R0m)}')
```

**Takeaway (C3, added after the first look):** the world total over the 18 columns falls **3.8%** ($21,414B → $20,593B). That moves every country's RCA in every column. Of the **23** M cells that flip in 2019, **18** are outside the aggregate column. The big movers have ch. 99 shares of 0.0–0.7% and are pushed across the threshold by a few hundredths. Congo's *Daily Instruments* goes 1.010 → 0.972, Moldova's *Transportation* 1.037 → 0.998, Zambia's *Precious Stones* 1.012 → 0.974, and Poland's aggregate 0.964 → 1.077. This is the near-threshold fragility 11-23 measured at HS4. Here the trigger is a change in the world denominator. Saudi Arabia is the only intended effect: *High-complexity Products* 1.727 → 0.253, but *Raw Manufacturing* 0.900 → 1.520, because its own total shrinks by 43.1%. That is why its rank only falls to 13 and not near 45.

## Caveats

- **ATLAS is a benchmark, not ground truth.** The Saudi Arabia diagnosis (C2b) rests on the raw HS6 data, not on ATLAS.
- **Small differences.** All 2019 bootstrap intervals for F4–F6 include 0. The failure under rule A is about *direction across years*, which the rule was designed to test, not about large effects.
- **E-style comparisons use R0 (rebuilt from raw), not the cleaned CSV** (Spearman 0.987–1.000 between them).
- **C3 and C2b were added after seeing B.** They explain the result. They don't change the rule-A verdict.
- Nominal values; 2023 suspect (D1). Saudi Arabia (2022) and Congo (2022–2023) are absent from the raw pulls, so the 2022 check year has no Saudi Arabia row.
- The peer dict is the one used by earlier runs (from the Spring 2026 PDFs). It was not re-read tonight.

## Helpful to know

- **The aggregate's double-counting moves every RCA, not just the aggregate.** Any change to what the aggregate contains changes the world total (3.8% here), which pushes near-threshold cells in *other* columns across RCA = 1. Count the flips outside the changed column before reading a mapping variant's rank changes as "caused" by the products moved.
- **Saudi Arabia reports oil as 999999** in 2019, 2020 and 2023, and as ch. 27 in 2021. Its grouped rank is sensible only in 2021. Any year-on-year Saudi Arabia comparison (the UAE report's peers) mixes two reporting conventions.
- **Qatar does the same in 2020 and 2022** (999999 = 85% and 77% of exports). Albania is ≥29% 999999 in every year.
- **11-24's F1 survives a new seed and both of tonight's exclusions.** It remains the only mapping variant that passes rule A.
